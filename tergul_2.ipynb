```python
import json
import ipywidgets as widgets
from IPython.display import display

FILE_PATH = "students.json"

# Read student data from JSON
def read_students():
    try:
        with open(FILE_PATH, "r") as file:
            data = json.load(file)
            return {student["first_name"]: student for student in data}
    except FileNotFoundError:
        print(f"❌ File not found: {FILE_PATH}")
        return {}
    except json.JSONDecodeError:
        print("⚠️ Error decoding JSON.")
        return {}

# Write student data to JSON
def write_students(students):
    try:
        with open(FILE_PATH, "w") as file:
            json.dump(list(students.values()), file, indent=4)
    except Exception as e:
        print(f"Error saving file: {e}")

# Load student data
students_data = read_students()

# Create dropdown to select student
student_dropdown = widgets.Dropdown(
    options=list(students_data.keys()),
    description="Student:"
)

# Create main text fields
first_name_text = widgets.Text(description="First Name:")
last_name_text = widgets.Text(description="Last Name:")
email_text = widgets.Text(description="Email:")
courses_text = widgets.Text(description="Courses:")
link_text = widgets.Text(description="Link:")

# Extra key/value input fields
extra_key_input = widgets.Text(description="New Field:")
extra_value_input = widgets.Text(description="Value:")

# Display selected student's details
def display_student_details(change):
    selected_name = student_dropdown.value
    if selected_name:
        student = students_data[selected_name]
        first_name_text.value = student.get("first_name", "")
        last_name_text.value = student.get("last_name", "")
        email_text.value = student.get("email", "")
        courses_text.value = student.get("courses", "")
        link_text.value = student.get("link", "")
        extra_key_input.value = ""
        extra_value_input.value = ""

# Update and save student data using .update()
def update_student_data(b):
    selected_name = student_dropdown.value
    if selected_name:
        updated_data = {
            "first_name": first_name_text.value,
            "last_name": last_name_text.value,
            "email": email_text.value,
            "courses": courses_text.value,
            "link": link_text.value
        }

        # Add extra field using .update() like your image
        key = extra_key_input.value.strip()
        value = extra_value_input.value.strip()
        if key and value:
            updated_data.update({key: value})

        students_data[selected_name] = updated_data
        write_students(students_data)
        print(f"✅ Data for {selected_name} updated successfully and saved!")

# Event listener
student_dropdown.observe(display_student_details, names="value")

# Update button
update_button = widgets.Button(description="Update Data", button_style="success")
update_button.on_click(update_student_data)

# Display everything
display(
    student_dropdown,
    first_name_text,
    last_name_text,
    email_text,
    courses_text,
    link_text,
    extra_key_input,
    extra_value_input,
    update_button
)

# Show first student on load
display_student_details(None)
```
