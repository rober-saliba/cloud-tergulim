```python
from firebase import firebase

# Initialize connection to your Firebase project
db = firebase.FirebaseApplication('https://test1-cd3f2-default-rtdb.firebaseio.com/', None)


def add_single_word(word):
    word = word.lower()
    current = db.get('/words/', word)
    if current is None:
        db.put('/words/', word, 1)
    else:
        db.put('/words/', word, current + 1)

def add_text(text):
    for word in text.lower().split():
        add_single_word(word)

def update_word_count(word, new_count):
    word = word.lower()
    db.put('/words/', word, new_count)

def delete_word(word):
    word = word.lower()
    db.delete('/words/', word)

def view_all_words():
    words = db.get('/words/', None)
    if words:
        for w, count in words.items():
            print(f"{w}: {count}")
    else:
        print("No words found.")

# ========== Menu ==========
while True:
    print("\nCommon Words Counter Menu:")
    print("1. Add single word")
    print("2. Add text for analysis")
    print("3. Update word count")
    print("4. Delete word")
    print("5. View all words")
    print("6. Exit")

    choice = input("Select an option (1-6): ").strip()

    if choice == '1':
        word = input("Enter word: ")
        add_single_word(word)

    elif choice == '2':
        text = input("Enter text: ")
        add_text(text)

    elif choice == '3':
        word = input("Enter word to update: ")
        new_count = int(input("Enter new count: "))
        update_word_count(word, new_count)

    elif choice == '4':
        word = input("Enter word to delete: ")
        delete_word(word)

    elif choice == '5':
        view_all_words()

    elif choice == '6':
        print("Exiting program...")
        break

    else:
        print("Invalid option. Please choose 1–6.")
```
